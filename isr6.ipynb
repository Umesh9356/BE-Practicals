```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt
```

```python
image_path = "input.webp"
image = Image.open(image_path).convert("RGB")

# Step 2: Convert image into a NumPy array
img_array = np.array(image)

# Step 3: Extract color features
red = img_array[:, :, 0]
green = img_array[:, :, 1]
blue = img_array[:, :, 2]

print("Image Feature Extraction")
print("-----------------------")
print("Image size:", image.size)
print("Red mean intensity:", round(np.mean(red), 2))
print("Green mean intensity:", round(np.mean(green), 2))
print("Blue mean intensity:", round(np.mean(blue), 2))

```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Color Image")
plt.axis("off")

# Step 5: Plot RGB histograms
plt.subplot(1, 2, 2)

plt.hist(red.ravel(), bins=256, range=(0, 256),
         color="red", alpha=0.5, label="Red")

plt.hist(green.ravel(), bins=256, range=(0, 256),
         color="green", alpha=0.5, label="Green")

plt.hist(blue.ravel(), bins=256, range=(0, 256),
         color="blue", alpha=0.5, label="Blue")

plt.title("RGB Color Histogram")
plt.xlabel("Pixel Intensity (0-255)")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()
plt.show()
```
