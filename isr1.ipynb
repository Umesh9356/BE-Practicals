```python
import re
from collections import Counter
```

```python
def simple_conflation(word):
    """
    A basic rule-based conflation (stemming) function 
    that strips common English suffixes.
    """
    word = word.lower()
    if word.endswith('ing') and len(word) > 4:
        word = word[:-3]
    elif word.endswith('ed') and len(word) > 3:
        word = word[:-2]
    elif word.endswith('es') and len(word) > 3:
        word = word[:-2]
    elif word.endswith('s') and len(word) > 2:
        word = word[:-1]
    return word
```

```python
def generate_document_representative(text):
    """
    Tokenizes text, removes punctuation/stop words, applies conflation,
    and returns a frequency dictionary (document representative).
    """
  
    stop_words = {'the', 'is', 'at', 'which', 'and', 'a', 'an', 'in', 'to', 'of'}
    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())
    
  
    processed_words = [simple_conflation(w) for w in words if w not in stop_words]
    
    doc_representative = Counter(processed_words)
    return doc_representative
```

```python
sample_text = """
Information retrieval systems are designed to store, organize, and retrieve 
large numbers of documents. Computing systems process user queries efficiently. Information
"""

doc_rep = generate_document_representative(sample_text)
print("--- Document Representative (Conflated Terms & Frequencies) ---")
for term, freq in doc_rep.items():
    print(f"'{term}': {freq}")
```
