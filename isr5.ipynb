```python
A = {"doc1.txt", "doc3.txt", "doc4.txt"}
Rq1 = {"doc1.txt", "doc2.txt", "doc3.txt"}
relevant_retrieved = A.intersection(Rq1)
if len(A) > 0:
    precision = len(relevant_retrieved) / len(A)
else:
    precision = 0
if len(Rq1) > 0:
    recall = len(relevant_retrieved) / len(Rq1)
else:
    recall = 0
if precision + recall > 0:
    f_measure = (2 * precision * recall) / (precision + recall)
else:
    f_measure = 0
if precision + recall > 0:
    e_measure = 1 - (
        (2 * precision * recall) / (precision + recall)
    )
else:
    e_measure = 1
print("Precision:", round(precision, 2))
print("Recall:", round(recall, 2))
print("F-measure:", round(f_measure, 2))
print("E-measure:", round(e_measure, 2))
```
