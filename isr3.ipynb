```python
import os
import re
from collections import defaultdict
```

```python
folder = r"C:\Users\umesh\Desktop\practical"
documents = {}

for filename in os.listdir(folder):
    if filename.endswith(".txt"):
        path = os.path.join(folder, filename)

        with open(path, "r", encoding="utf-8") as file:
            documents[filename] = file.read()

# Step 2: Create an inverted index
inverted_index = defaultdict(set)

for filename, content in documents.items():
    words = re.findall(r"\b[a-zA-Z]+\b", content.lower())

    for word in words:
        inverted_index[word].add(filename)

# Step 3: Display the inverted index
print("\nINVERTED INDEX")
for word in sorted(inverted_index):
    print(f"{word}: {sorted(inverted_index[word])}")

# Step 4: Search for documents
query = input("\nEnter search keywords: ").lower()
query_words = re.findall(r"\b[a-zA-Z]+\b", query)

if not query_words:
    print("Please enter valid search keywords.")
else:
    # AND retrieval: all query words must be present
    result_sets = [
        inverted_index.get(word, set())
        for word in query_words
    ]

    results = set.intersection(*result_sets)

    # Step 5: Display matching documents
    print("\nRETRIEVED DOCUMENTS")

    if results:
        for filename in sorted(results):
            print(f"\nFile: {filename}")
            print(documents[filename])
    else:
        print("No matching documents found.")
```
