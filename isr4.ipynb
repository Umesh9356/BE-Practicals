```python
A = {"doc1.txt", "doc3.txt", "doc4.txt"}
Rq1 = {"doc1.txt", "doc2.txt", "doc3.txt"}
relevant_retrieved = A.intersection(Rq1)
if len(A) > 0:
    precision = len(relevant_retrieved) / len(A)
else:
    precision = 0

if len(Rq1) > 0:
    recall = len(relevant_retrieved) / len(Rq1)
else:
    recall = 0
print("Retrieved Documents (A):", sorted(A))
print("Relevant Documents (Rq1):", sorted(Rq1))
print("Relevant Retrieved Documents:", sorted(relevant_retrieved))
print("Precision:", round(precision, 2))
print("Precision (%):", round(precision * 100, 2), "%")
print("Recall:", round(recall, 2))
print("Recall (%):", round(recall * 100, 2), "%")
```
